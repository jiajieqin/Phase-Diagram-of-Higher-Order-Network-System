```python
from pypdf import PdfReader, PdfWriter, Transformation

pdf1 = PdfReader("./SIS/SIS_1017.pdf")
pdf3 = PdfReader("./GRN/grn_1017.pdf")
pdf5 = PdfReader("./Neuronal/neuronal_1017.pdf")

pdf2 = PdfReader("./hysteresis/sis/SIS_hysteresis.pdf")
pdf4 = PdfReader("./hysteresis/grn/GRN_hysteresis.pdf")
pdf6 = PdfReader("./hysteresis/neuronal/Neuronal_hysteresis.pdf")

page1 = pdf1.pages[0]
page2 = pdf2.pages[0]
page3 = pdf3.pages[0]

page4 = pdf4.pages[0]
page5 = pdf5.pages[0]
page6 = pdf6.pages[0]

w1 = float(page1.mediabox.width)
h1 = float(page1.mediabox.height)
w2 = float(page2.mediabox.width)
h2 = float(page2.mediabox.height)
w3 = float(page3.mediabox.width)
h3 = float(page3.mediabox.height)

w4 = float(page4.mediabox.width)
h4 = float(page4.mediabox.height)
w5 = float(page5.mediabox.width)
h5 = float(page5.mediabox.height)
w6 = float(page6.mediabox.width)
h6 = float(page6.mediabox.height)

new_width = w1 + w2 -10
new_height = h1 + h3 + h5 - 30 #max(h1, h2, h3)

y_top = max(h5, h6) - 20
y_top1 = max(h5, h6) + max(h3, h4) - 25
writer = PdfWriter()
new_page = writer.add_blank_page(width=new_width, height=new_height)

new_page.merge_transformed_page(
    page1,
    Transformation().translate(tx=0, ty=y_top1)
)

new_page.merge_transformed_page(
    page2,
    Transformation().translate(tx=w1-7, ty=y_top1)
)

new_page.merge_transformed_page(
    page3,
    Transformation().translate(tx=0, ty=y_top)
)

new_page.merge_transformed_page(
    page4,
    Transformation().translate(tx=w1-7, ty=y_top)
)

new_page.merge_transformed_page(
    page5,
    Transformation().translate(tx=0, ty=-10)
)

new_page.merge_transformed_page(
    page6,
    Transformation().translate(tx=w1 -5 , ty=-10)
)

with open("result_all_dynamics2.pdf", "wb") as f:
    writer.write(f)
```
