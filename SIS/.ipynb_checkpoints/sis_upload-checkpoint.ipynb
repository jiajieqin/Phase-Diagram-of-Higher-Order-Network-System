```python
structure = hdf5storage.loadmat('/home/qinjiajie/jupyter_code/Critical_transition_HOI/Empirical_data/after_operation/sp_high_edges_3000.mat')
```

```python
A = structure['A']
B = structure['B'][0]
g1 = 0.01
```

```python
size_B = len(B)
B_ijk = np.zeros((size_B, size_B, size_B))
for i in range(size_B):
    for j in range(len(B[i])):
        B_temp = B[i][j]
        B_ijk[i][B_temp[0]][B_temp[1]] = 1
        B_ijk[i][B_temp[1]][B_temp[0]] = 1
B_ij = np.zeros((size_B, size_B))
for i in range(size_B):
    for j in range(size_B):
        B_ij[i][j] = np.sum(B_ijk[i,j,:])
B_ij = 0.5*B_ij
```

```python
lamda = np.linalg.eig(A)
index = np.argmax(lamda[0])
rho = np.real(lamda[0][index])
rho
```

```python
zz = np.load('zz.npy')
```

```python
result_large_origin = hdf5storage.loadmat('true_collapse.mat')
true_collapse_tipping = result_large_origin['true_collapse']
result_large_origin = hdf5storage.loadmat('true_collapse1.mat')
true_collapse_tipping1 = result_large_origin['zz']
true_collapse_tipping = np.vstack([true_collapse_tipping1, true_collapse_tipping])
```

```python
result_large_2d = hdf5storage.loadmat('theory_collapse.mat')
theory_collapse_tipping = result_large_2d['theory_collapse']
result_large_2d = hdf5storage.loadmat('theory_collapse1.mat')
theory_collapse_tipping1 = result_large_2d['zz']
theory_collapse_tipping = np.vstack([theory_collapse_tipping1, theory_collapse_tipping])
```

```python
def f2_threshold(paras1,paras2,g1):
    alpha,beta,gamma,alpha1,beta1,gamma1,rho = paras1
    # d1dt = f*x[0] + g1*alpha*(1-beta*x[0])*gamma*x[0] + g2*alpha1*(1-beta1*x[1])*gamma1*x[1]*rho*x[1]
    
    alpha_2,beta_2,gamma_2,alpha1_2,beta1_2,gamma1_2,rho_2 = paras2

    beta_ = 0.5*(beta+beta_2)
    beta1_ = 0.5*(beta1 + beta1_2)
    
    A = alpha_2*gamma_2/alpha
    B = alpha1_2/(alpha1*gamma1*rho)

    c1 = A
    c2 = beta_*(B-A)
    z = (3*beta1_*c1 + np.sqrt((3*beta1_*c1)**2 + 32*beta1_*c2))/(8*beta1_)
    g2 = (g1*alpha*beta_/(alpha1*gamma1*rho))*(4*z-3*c1)/(z**2 * (2*z - c1))
    return g2
```

```python
f2_crit = f2_threshold(zz[0][:7],zz[1][:7],g1)
```

```python
from matplotlib.ticker import MaxNLocator, FuncFormatter
with plt.style.context(['science', 'no-latex']): 
    fig, ax = plt.subplots(nrows=1,ncols=1,figsize=(5, 4))
    bwith = 0.2  # 边框宽度设置为2
    ax.spines['bottom'].set_linewidth(bwith)
    ax.spines['left'].set_linewidth(bwith)
    ax.spines['top'].set_linewidth(bwith)
    ax.spines['right'].set_linewidth(bwith)

    ax.fill_between(np.arange(0,0.051,0.001),
                    0*np.ones(len(np.arange(0,0.051,0.001))),
                    0.01*rho*np.ones(len(np.arange(0,0.051,0.001))),
                    facecolor='#D4E6F2',label='Active region (simulation)')
    
    ax.fill_betweenx(-true_collapse_tipping[:,1],
                     true_collapse_tipping[:,0],
                    0.05*np.ones(len(true_collapse_tipping[:,0],)),
                    facecolor='#D4F2D5',label='Sensitive region (simulation)')

    ax.fill_betweenx(-true_collapse_tipping[:,1],
                     true_collapse_tipping[:,0],
                    0*np.ones(len(true_collapse_tipping[:,0],)),
                    facecolor='#FDDBD9',label='Inactive region (simulation)')
    ax.plot(theory_collapse_tipping[:,0],-theory_collapse_tipping[:,1],label='Boundaries (theory)',c='#808080',linewidth =1.5,linestyle='--')

    ax.axhline(y=g1*rho,c='#808080',linewidth =1.5,linestyle='--') ##d170a8
    
    ax.set_ylim([0.0001,1])
    ax.set_xlim([0,0.05])
    ax.scatter(f2_crit, g1*rho, zorder=4, marker='o', s=20,
            facecolors='k', edgecolors='k', linewidths=3, alpha=0.7)
    # ax.scatter(0.045, 0.945, s=10,zorder=5)
    ax.set_xlabel(r'$g_2$',fontsize=20)
    ax.set_ylabel(r'$s$',fontsize=20)
    # ax.tick_params(width=0.1)
    ax.yaxis.set_major_locator(MaxNLocator(5))
    ax.xaxis.set_major_locator(MaxNLocator(5))
    ax.set_title('Epidemic spreading',fontsize=20, pad=10)
    ax.text(-0.2,1.1,'(a)',fontsize=15,transform=ax.transAxes)
    ax.legend(frameon=True,facecolor='white',fontsize=9, loc='upper left')
    ax.tick_params(axis='both', labelsize=13, pad=7)
    ax.xaxis.set_major_formatter(FuncFormatter(lambda x, pos: '0' if abs(x) < 1e-12 else f'{x:.2f}'))
plt.tight_layout()
plt.savefig("SIS_0902.pdf", format='pdf', dpi=300)
plt.show()
```
